```python
!pip install -U langchain langchain-community pypdf langchain_text_splitters sentence-transformers
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import DirectoryLoader
```

```python
# Extract text from PDF files.
def load_pdf_files(directory):
    loader = DirectoryLoader(
        directory,
        glob = "*.pdf",
        loader_cls = PyPDFLoader
    )
    documents = loader.load()
    return documents
```

```python
# Load the data.
extracted_text = load_pdf_files("/content/")
extracted_text
```

```python
len(extracted_text)
```

```python
from typing import List
from langchain_core.documents import Document

def filter_to_minimal_docs(docs: List[Document]) -> List[Document]:
    minimal_docs = []
    for doc in docs:
        minimal_doc = Document(
            page_content=doc.page_content,
            metadata={"source": doc.metadata.get("source", "")}
        )
        minimal_docs.append(minimal_doc)
    return minimal_docs
```

```python
# Split the documents into smaller chunks.

def text_split(minimal_docs):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
    )
    split_docs = text_splitter.split_documents(minimal_docs)
    return split_docs
```

```python
#  call the text split function
minimal_docs = filter_to_minimal_docs(extracted_text)
len(minimal_docs)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

def download_embeddings():
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )
    return embeddings


embedding = download_embeddings()
```

```python
# testing our embedding model.
embedding.embed_query("What acan you do ?")
```

```python
print(f"Vecotr lenght: {len(embedding.embed_query("What acan you do ?"))}")
```
