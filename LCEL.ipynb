```python
from langchain_openai import ChatOpenAI

chatModel = ChatOpenAI(
    # model="gpt-3.5-turbo",
    openai_api_key=open_ai,
)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "tell me a curious fact about {cameroon}"
)

chain = prompt | chatModel | StrOutputParser()
```

```python
# chain.invoke({"cameroon": "what is happening now in Cameroon"})
chain.invoke("what is happening now in Cameroon")
```

```python
for stream in chain.stream({"cameroon": "what is happening now in Cameroon"}):
    print(stream)
```
