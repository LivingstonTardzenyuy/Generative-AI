```python
from google.colab import userdata
grokAI = userdata.get('GROK_api')
openAI = userdata.get('OPENAI_API_KEY')
tavily = userdata.get('tavily')
```

```python
from langchain_openai import ChatOpenAI

chatgpt_llm = ChatOpenAI(
    model = "gpt-3.5-turbo",
    openai_api_key = openAI
)
```

```python
from langchain_community import ChatGrok

grok_llm = ChatGrok(
    model = "grok-1",
    grok_api_key = grokAI
)
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults

search = TavilySearchResults(
    tavily_api_key = tavily,
    max_results = 2
)

search.invoke("what schools are in University of Bamenda  ?")
```

```python
tools = [search]
```

```python
from langchain.agents import create_react_agent

agent_executor = create_react_agent(
    chatgpt_llm, tools
)
```

```python
from langchain_core.messages import HumanMessage

response = agent_executor.invoke(
    {
        "messages": [HumanMessage(
            content = "How was the university of bamenda created, can i enroll"
        )]
    }
)

response['messages']
```

```python
from langchain_core.messages import HumanMessage

response = agent_executor.invoke(
    {
        "messages": [HumanMessage(
            content = "What is Python"
        )]
    }
)

response['messages']
```

```python
for chunk in agent_executor.stream(
    {"messages": [HumanMessage(content="when will the graduation of the University of bamenda ")]}
):

  print(chunk)
  print("-----")
```

## Adding Memory

Adding  memory in LangGraph is very similar to in Langchain

```python
from langgraph.checkpoint.memory import MemorySaver

# config = MemorySaver(
#     "configurable": {"thread_id": "001"}
# )
```

```python
from langgraph.checkpoint.memory import MemorySaver
from langchain.agents import create_react_agent

memory = MemorySaver()

agent_executor = create_react_agent(
    chatgpt_llm, tools
)

config = {
    "configurable": {
        "thread_id": "001",
        "checkpoint_saver": memory
    }
}
```

```python
for chunk in agent_executor.stream(
    {"messages": [HumanMessage(content="when will the graduation of the University of bamenda ")]}
):
  print(chunk)
  print('----')
```
