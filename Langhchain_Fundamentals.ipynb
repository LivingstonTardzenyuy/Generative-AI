```python
from google.colab import userdata
import os

# Access the OpenAI API key from Colab Secrets
openai_api_key = userdata.get("OPENAI_API_KEY")

if openai_api_key:
    print("OpenAI API key loaded successfully from Colab Secrets.")
else:
    print("OpenAI API key not found in Colab Secrets. Please add it using the '🔑 Secrets' icon on the left sidebar.")
```

```python
# Completion model.
from langchain_openai import OpenAI

llmModel = OpenAI(openai_api_key=openai_api_key)
```

```python
response = llmModel.invoke("What is happening in Cameroon of recent ?")
print(response)
```

```python
# Initializing the temp parameter.
creativeModel = OpenAI(
    openai_api_key=openai_api_key,
    temperature=0.9
)
response = creativeModel.invoke("What is happening in Cameroon of recent ?")
print(response)
```

```python
# Streaming.
for chunk in creativeModel.stream("Why do parrots have colorful feathers?"):
    print(chunk, end="|", flush=True)
```
