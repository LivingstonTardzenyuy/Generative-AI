## RunnablePassthrough

It does not do anything to the input data.
Let's see it in a very simple example: A chain with just RunnablePassthrough() will output the original input without any modifications

```python
from langchain_core.runnables import RunnablePassthrough

# create runnable of runnable
chain = RunnablePassthrough()
```

```python
chain.invoke("Kongnyuy")
```

##RunnableLambda
To use a custom function inside a LCEL chain we need to wrap it up with RunnableLambda


```python
def lamso_lastname(name: str) -> str:
  return f"{name}yuy"
```

```python
from langchain_core.runnables import RunnableLambda

chain = RunnablePassthrough() | RunnableLambda(lamso_lastname)
```

```python
chain.invoke("Livingston")
```
